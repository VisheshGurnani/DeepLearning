# Loss Functions: Mean Squared Error vs Cross-Entropy Loss

### Objective
Understand how neural networks measure prediction errors by calculating and comparing **Mean Squared Error (MSE)** and **Cross-Entropy Loss** for a simple classification problem.

## Task 1: Actual Labels and Predicted Outputs

We consider a binary classification problem where:

- `1` = Positive class
- `0` = Negative class

The actual class labels are compared with the model's predicted probabilities.

```python
import numpy as np

# Actual class labels
y_actual = np.array([1, 0, 1, 1, 0])

# Predicted probabilities for the positive class
y_pred = np.array([0.9, 0.2, 0.8, 0.7, 0.3])

print("Actual labels:     ", y_actual)
print("Predicted outputs: ", y_pred)
```

## Task 2: Calculate Mean Squared Error

Mean Squared Error measures the average squared difference between actual values and predicted values.

\[
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
\]

```python
# Calculate Mean Squared Error manually
mse = np.mean((y_actual - y_pred) ** 2)

print(f"Mean Squared Error (MSE): {mse:.4f}")
```

## Calculate Cross-Entropy Loss

For binary classification, Binary Cross-Entropy is:

\[
L = -\frac{1}{n}\sum_{i=1}^{n}
[y_i\log(\hat{y}_i) + (1-y_i)\log(1-\hat{y}_i)]
\]

A small epsilon is used to avoid taking the logarithm of zero.

```python
# Calculate Binary Cross-Entropy Loss manually
epsilon = 1e-15
y_pred_clipped = np.clip(y_pred, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    y_actual * np.log(y_pred_clipped) +
    (1 - y_actual) * np.log(1 - y_pred_clipped)
)

print(f"Cross-Entropy Loss: {cross_entropy:.4f}")
```

## Task 3: Compare the Results

The two loss functions measure prediction error differently.

```python
print("Loss Function Comparison")
print("-" * 30)
print(f"MSE                : {mse:.4f}")
print(f"Cross-Entropy Loss : {cross_entropy:.4f}")
```

### Explanation

**Mean Squared Error (MSE):**
- Measures the squared numerical difference between actual and predicted values.
- Commonly used for **regression problems**, where the target is a continuous numerical value.
- It can also be used for classification in some situations, but it is generally not the preferred loss for probabilistic classification.

**Cross-Entropy Loss:**
- Designed for classification problems.
- Measures how well predicted probabilities match the actual class labels.
- Strongly penalizes confident but incorrect predictions.
- Commonly used with **sigmoid outputs for binary classification** and **softmax outputs for multiclass classification**.

### Conclusion

For this binary classification example, **Cross-Entropy Loss is generally more appropriate than MSE** because it is specifically designed to evaluate probabilistic classification predictions.

The calculated loss values show that both functions can quantify prediction error, but they interpret the error differently. The choice of loss function should depend on the type of machine learning problem and the model's output.
