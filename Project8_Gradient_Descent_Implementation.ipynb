# Project 8: Gradient Descent Implementation

Implement Gradient Descent and observe how model parameters update while prediction error decreases.

## Task 1: Dataset and Initial Parameters

We use a simple linear relationship, `y = 2x + 1`, and a model `y_pred = wx + b`.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simple dataset
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

# Initialize parameters
w = 0.0
b = 0.0

learning_rate = 0.01
iterations = 100

print("Initial weight (w):", w)
print("Initial bias (b):", b)
print("Learning rate:", learning_rate)
print("Iterations:", iterations)
```

## Task 2: Implement Gradient Descent

At each iteration we calculate predictions, Mean Squared Error, gradients, and then update `w` and `b`.

```python
loss_history = []
n = len(X)

for i in range(iterations):
    # Predictions
    y_pred = w * X + b

    # Mean Squared Error
    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    # Gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Parameter updates
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Final weight (w):", round(w, 4))
print("Final bias (b):", round(b, 4))
print("Final loss:", round(loss_history[-1], 4))
```

## Task 3: Analyze Loss Reduction

The graph below shows how the Mean Squared Error decreases during training.

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, iterations + 1), loss_history)
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.title("Loss Reduction During Gradient Descent")
plt.grid(True)
plt.show()
```

## Visualize the Learned Model

Compare the original data with the predictions produced by the trained model.

```python
final_predictions = w * X + b

plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual Data")
plt.plot(X, final_predictions, label="Gradient Descent Model")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Actual Data vs Learned Model")
plt.legend()
plt.grid(True)
plt.show()

print("Actual values:    ", y)
print("Predicted values: ", np.round(final_predictions, 2))
```

## Conclusion

Gradient Descent repeatedly updates model parameters in the direction that reduces the loss.

**Expected Result:**
- The loss decreases over multiple iterations.
- The weight approaches **2**.
- The bias approaches **1**.
- Predictions become closer to the actual values.

This demonstrates the basic optimization process used to train machine learning and neural network models.
