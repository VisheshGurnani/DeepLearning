```python
import numpy as np
import matplotlib.pyplot as plt

#Task 1: Define Activation Function
def sigmoid(x):
  return 1 / (1+np.exp(-x))

#ReLU Finction
def relu(x):
  return np.maximum(0,x)

#Tanh Function
def tanh(x):
  return np.tanh(x)
```

```python
#Task 2: Generate Outputs

# Input values
x_values=np.array([-5,-3,-1,0,1,3,5])

#Apply Activation Functions
sigmoid_outputs=sigmoid(x_values)
relu_outputs=relu(x_values)
tanh_outputs=tanh(x_values)

#Display Outputs
print("Input Values:",x_values)
print("\nSigmoid Outputs:",sigmoid_outputs)
print("\nReLU Outputs:",relu_outputs)
print("\nTanh Outputs:",tanh_outputs)
```

```python
#Task 3: Visualize and Compare

#Generate continuous input values for plotting
x = np.linspace(-5, 5, 100)

#Calculate outputs for each activation function
sigmoid_y = sigmoid(x)
relu_y = relu(x)
tanh_y = tanh(x)

#Plot the activation functions
plt.figure(figsize=(10, 6))
plt.plot(x, sigmoid_y, label='Sigmoid', color='blue')
plt.plot(x, relu_y, label='ReLU', color='green')
plt.plot(x, tanh_y, label='Tanh', color='red')
plt.title('Activation Functions Comparison')
plt.xlabel('Input')
plt.ylabel('Output')
plt.grid(True)
```
